# Lissajous Pseudo-Knots Self-Intersections

Let's define a couple of functions to plot different types of knots and calculate their arc-lenghts.

```python
# Arc-length function

from scipy.integrate import quad
import numpy as np

def arclength(nx,ny,nz, φx,φy,φz, ktype=str('Lissajous'), q=1):

  if ktype == str('Lissajous'): # Lissajous knot
    a = lambda t: np.sqrt((nx*np.sin(nx*t + φx))**2 + (ny*np.sin(ny*t + φy))**2 + (nz*np.sin(nz*t + φz))**2)

  elif ktype == str('fib'): # Fibonacci knot
    φx = φy = φz = 0.5
    a = lambda t: np.sqrt((nx*np.sin(nx*t + φx))**2 + (ny*np.sin(ny*t + φy))**2 +
                          (0.5*nz*np.sin(nz*t + φz) + 0.5*ny*np.cos(ny*t + φy))**2)

  elif ktype == str('trefoil'): # Trefoil knot
    a = lambda t: np.sqrt(  (  np.cos(t) + 4*np.cos(2*t))**2
                          + (- np.sin(t) + 4*np.sin(2*t))**2
                          + (3*np.cos(3*t))**2 )

  #The (2,3)-torus knot is also a trefoil knot.
  elif ktype == str('torus'): # Torus knot
    a = lambda t: np.sqrt(  (np.sin(t)*(1 + 0.5*np.cos(q*t)) + 0.5*q*np.sin(q*t)*np.cos(t))**2
                          + (np.cos(t)*(1 + 0.5*np.cos(q*t)) - 0.5*q*np.sin(q*t)*np.sin(t))**2
                          + (0.5*q*np.cos(q*t))**2 )

  arc = ( round( quad( a, 0, (1/2)*np.pi )[0], 2 ) + round( quad( a, (1/2)*np.pi, np.pi)[0], 2 )
      + round( quad( a, np.pi, (3/2)*np.pi)[0], 2 ) + round( quad( a, (3/2)*np.pi, 2*np.pi)[0], 2) )

  # print('The length of the', ktype, 'knot is:')
  return arc
```

```python
# Plotting function for the Lissajous Knots

import numpy as np
import plotly.graph_objects as go

def Liss(nx,ny,nz, φx=0,φy=0,φz=0, xx=2.4,yy=0,zz=0, div=1, ktype=str('liss'), q=1, save_path=None, circle=None): #(xx,yy,zz) is the camera position
  # xx=2.4,yy=0.001,zz=0.07 for better axis presentation but deviated from symmetric view-point

# def Liss(a,b, nx,ny,nz, φx=0,φy=0,φz=0, xx=-2.5,yy=0,zz=0, div=1, ktype=str('liss'), q=1):
#     r = [a,b]
  
  # 2,5,(even x odd): (2,5,2), (2,5,6), (2,5,14), etc
  # 2,5,(even) and 2,5,(multiple of 5) give additional intersections
  if nx==2 and ny==5 and φx==(1/2)*np.pi:
    r = []
    for i in range(1,ny):
      r.append(np.pi*((1/2)*(ny*nz) - nz*i)/(ny*nz))
      r.append(np.pi*((3/2)*(ny*nz) - nz*i)/(ny*nz))

  # pi/3 Symmetry
  # Liss(3,4,5, 0,0,np.pi/3)
  elif nx==3 and ny==4 and nz==5 and φz==(1/3)*np.pi:
    # r = [(1/12)*np.pi,(7/12)*np.pi, -(2/12)*np.pi,(10/12)*np.pi, -(5/12)*np.pi,(13/12)*np.pi]
    r = []
    for i in range(nx):
      r.append(np.pi*(1 - 3*i)/(nx*ny))
      r.append(np.pi*(7 + 3*i)/(nx*ny))

  # Liss(3,4,7, 0,0,np.pi/3)
  elif nx==3 and ny==4 and nz==7 and φz==(1/3)*np.pi:
    # r = [(11/12)*np.pi,(5/12)*np.pi, (14/12)*np.pi,(2/12)*np.pi, (17/12)*np.pi,-(1/12)*np.pi]
    r = []
    for i in range(nx):
      r.append(np.pi*(11 + 3*i)/(nx*ny))
      r.append(np.pi*( 5 - 3*i)/(nx*ny))

  # elif ny % 2 == 0:
    # for i in range(nx):
    #   r.append( ( (ny+nz + 3*i) /(nx*ny) )*np.pi )
    #   print((ny+nz + 3*i))
    #   r.append( ( (nx*ny-nz - 3*i) /(nx*ny) )*np.pi )
    #   print((nx*ny-nz - 3*i))
      # r.append( ( (ny+nz + 3*i) /(nx*ny) )*np.pi )
      # r.append( ( (nx*ny-nz - 3*i) /(nx*ny) )*np.pi )

  elif nx==4 and ny==5 and nz==15 and φz==(4/5)*np.pi:
    r = []
        #   (18/20)*np.pi,  (2/20)*np.pi, # horizontal
        #  -(18/20)*np.pi, -(2/20)*np.pi,
        #   (14/20)*np.pi,  (6/20)*np.pi,
        #  -(14/20)*np.pi, -(6/20)*np.pi,
        #   (9/20)*np.pi,   (1/20)*np.pi, # diagonal short
        #  -(9/20)*np.pi,  -(1/20)*np.pi,
        #   (19/20)*np.pi,  (11/20)*np.pi,
        #  -(19/20)*np.pi, -(11/20)*np.pi,
        #  -(13/20)*np.pi,  (3/20)*np.pi, # diagonal long
        #   (13/20)*np.pi, -(3/20)*np.pi,
        #  -(17/20)*np.pi,  (7/20)*np.pi,
        #   (17/20)*np.pi, -(7/20)*np.pi]
    for i in range(2):
      # Horizontal symmetries
      r.append(  (( 4*i + 2 )/(nx*ny))*np.pi )
      r.append(  ((-4*i + 18)/(nx*ny))*np.pi )
      r.append( -(( 4*i + 2 )/(nx*ny))*np.pi )
      r.append( -((-4*i + 18)/(nx*ny))*np.pi )
      # Short Diagonal symmetries
      r.append(  ((10*i + 1)/(nx*ny))*np.pi )
      r.append(  ((10*i + 9)/(nx*ny))*np.pi )
      r.append( -((10*i + 1)/(nx*ny))*np.pi )
      r.append( -((10*i + 9)/(nx*ny))*np.pi )
      # Long Diagonal symmetries
      r.append(  ((4*i + 3 )/(nx*ny))*np.pi )
      r.append( -((4*i + 13)/(nx*ny))*np.pi )
      r.append( -((4*i + 3 )/(nx*ny))*np.pi )
      r.append(  ((4*i + 13)/(nx*ny))*np.pi )

  else:
    r = []

  # r = [-(1/6)*np.pi, (5/6)*np.pi,  (1/6)*np.pi, -(5/6)*np.pi]
  # r = [-(1/6)*np.pi, np.pi-(1/6)*np.pi]

  # r = [0*np.pi/10,10*np.pi/10, 2*np.pi/10,8*np.pi/10, 3*np.pi/10,7*np.pi/10, 4*np.pi/10,6*np.pi/10, 5*np.pi/10,-5*np.pi/10, -2*np.pi/10,-8*np.pi/10, -3*np.pi/10,-7*np.pi/10, -4*np.pi/10,-6*np.pi/10]

  # r = [1*np.pi/10,9*np.pi/10, 3*np.pi/10,7*np.pi/10, -3*np.pi/10,-7*np.pi/10, -1*np.pi/10,-9*np.pi/10]


  # Vertical Symmetry
  if φx==0 and φy==0:
    div = nz
  elif φx==0 and φz==0:
    div = ny
  elif φy==0 and φz==0:
    div = nx
  if φx not in {0, np.pi} or φy not in {0, np.pi} or φz not in {0, np.pi}: # Removes dots at knot collapse
    for i in range(1,div):
      r.append((i/div)*np.pi)
      r.append(-(i/div)*np.pi)
      
  # Adds multiple dots at knot collapse
  # elif φx in {0, np.pi} or φy in {0, np.pi} or φz in {0, np.pi}:
  #   tot = 200
  #   for i in range(tot):
  #     r.append((i/tot)*np.pi)
  #     r.append( np.pi-(i/tot)*np.pi) # horizontal (use tot/2)
  #     r.append(-(i/tot)*np.pi)
  #     r.append(-np.pi+(i/tot)*np.pi) # horizontal

  # Horizontal Symmetry
  if nx % 2 == 0:
    if φx==0 and φy==0 and (φz==(1/2)*np.pi or φz==(3/2)*np.pi):
      for i in range(1,ny,2):
        r.append( i/(2*ny)*np.pi )
        r.append( (2*ny-i)/(2*ny)*np.pi )
        r.append( -i/(2*ny)*np.pi )
        r.append( -(2*ny-i)/(2*ny)*np.pi )

    if φz==0 and φx==0 and (φy==(1/2)*np.pi or φy==(3/2)*np.pi):
      for i in range(1,nz,2):
        r.append( i/(2*nz)*np.pi )
        r.append( (2*nz-i)/(2*nz)*np.pi )
        r.append( -i/(2*nz)*np.pi )
        r.append( -(2*nz-i)/(2*nz)*np.pi )

  if ny % 2 == 0:
    if φy==0 and φz==0 and (φx==(1/2)*np.pi or φx==(3/2)*np.pi):
      for i in range(1,nz,2):
        r.append( i/(2*nz)*np.pi )
        r.append( (2*nz-i)/(2*nz)*np.pi )
        r.append( -i/(2*nz)*np.pi )
        r.append( -(2*nz-i)/(2*nz)*np.pi )

  # Diagonal Symmetry
  if nx % 2 == 0:
    if φy==(1/5)*np.pi or φy==(6/5)*np.pi:
      # for i in range(nx):
      #   r.append( ((nx/2 + i*nz)/(nx*nz))*np.pi )
      #   r.append( ((-1)**i)*((nx*nz - (nx/2 + 1))/(nx*nz))*np.pi )
      r.append(  ((nx/2)/(nx*nz))*np.pi )
      r.append( -((nx*nz - (nx/2))/(nx*nz))*np.pi )
      r.append(  ((nx/2+nz)/(nx*nz))*np.pi )
      r.append(  ((nx*nz - (ny-0))/(nx*nz))*np.pi )
      r.append( -((nx*nz-(nx/2+nz))/(nx*nz))*np.pi )
      r.append( -((ny-0)/(nx*nz))*np.pi )

    if φy==(4/5)*np.pi or φy==(9/5)*np.pi:
      r.append( -1/(2*nz)*np.pi )
      r.append( (2*nz-1)/(2*nz)*np.pi )

    if φy==(2/5)*np.pi or φy==(7/5)*np.pi:
      r.append( -3/(2*nz)*np.pi )
      r.append( (2*nz-3)/(2*nz)*np.pi )

    if φy==(3/5)*np.pi or φy==(8/5)*np.pi:
      r.append( 3/(2*nz)*np.pi )
      r.append( -(2*nz-3)/(2*nz)*np.pi )

    # 4,5,nz, 0,0,(m/5)*np.pi
    if nx==4 and ny==5 and nz==15:
      None
    elif nx==4 and ny==5:
      if nz==7 or nz==17 or nz==27 or nz==37:
        extra = 2
      elif nz==9 or nz==19 or nz==29:
        extra = 6
      elif nz==11 or nz==21 or nz==1:
        extra = 14
      elif nz==13 or nz==23 or nz==3:
        extra = 18

      if φz==(4/5)*np.pi or φz==(9/5)*np.pi:
        r.append( (nx*ny+extra)/(nx*ny)*np.pi ) # nz=7 -> 22, nz=9 -> 26
        r.append( (nx*ny+extra)/(nx*ny)*np.pi + np.pi)
        r.append( (nx*ny+extra - ny)/(nx*ny)*np.pi )
        r.append( (nx*ny+extra + ny)/(nx*ny)*np.pi + np.pi)
        r.append( (nx*ny+extra + ny)/(nx*ny)*np.pi )
        r.append( (nx*ny+extra - ny)/(nx*ny)*np.pi + np.pi)

      if φz==(1/5)*np.pi or φz==(6/5)*np.pi:
        r.append( -(nx*ny+2)/(nx*ny)*np.pi )
        r.append( -(nx*ny+2)/(nx*ny)*np.pi + np.pi)
        r.append( -(nx*ny+2 - ny)/(nx*ny)*np.pi )
        r.append( -(nx*ny+2 + ny)/(nx*ny)*np.pi + np.pi)
        r.append( -(nx*ny+2 + ny)/(nx*ny)*np.pi )
        r.append( -(nx*ny+2 - ny)/(nx*ny)*np.pi + np.pi)

      if φz==(3/5)*np.pi or φz==(8/5)*np.pi:
        r.append( (2*nz)/(nx*ny)*np.pi )
        r.append( (2*nz)/(nx*ny)*np.pi + np.pi)
        r.append( (2*nz-ny)/(nx*ny)*np.pi )
        r.append(-(2*nz+nz)/(nx*ny)*np.pi + np.pi)
        r.append( (2*nz+ny)/(nx*ny)*np.pi )
        r.append( (2*nz-ny)/(nx*ny)*np.pi + np.pi)

      if φz==(2/5)*np.pi or φz==(7/5)*np.pi:
        r.append(-(2*nz)/(nx*ny)*np.pi )
        r.append(-(2*nz)/(nx*ny)*np.pi + np.pi)
        r.append(-(2*nz-ny)/(nx*ny)*np.pi )
        r.append( (2*nz+nz)/(nx*ny)*np.pi + np.pi)
        r.append(-(2*nz+ny)/(nx*ny)*np.pi )
        r.append(-(2*nz-ny)/(nx*ny)*np.pi + np.pi)
    

  # Odd nx
  # if nx % 2 == 1:
  #   if φy == 0.2*np.pi:
  #     r.append( -3/(2*nz)*np.pi )
  #     r.append( (2*nz-3)/(2*nz)*np.pi )

  if ktype == str('liss'): # Lissajous knot
    arc = arclength(nx,ny,nz, φx,φy,φz)
    intarc = int(round(arc,0))
    npts = 20*intarc
    t = np.linspace(0, 2*np.pi, npts)
    x,y,z = np.cos(nx*t + φx), np.cos(ny*t + φy), np.cos(nz*t + φz) #Knot Points
    

  if ktype == str('fib'): # Fibonacci knot
    arc = arclength(nx,ny,nz, φx,φy,φz, ktype)
    intarc = int(round(arc,0))
    npts = 20*intarc
    t = np.linspace(0, 2*np.pi, npts)
    x,y = np.cos(nx*t + φx), np.cos(ny*t + φy)
    z = 0.5*np.cos(nz*t + φz) + 0.5*np.sin(ny*t + φy)
    r = np.array(r)
    a,b = np.cos(nx*r + φx), np.cos(ny*r + φy) # Fibonacci Nodes
    c = 0.5*np.cos(nz*r + φz) + 0.5*np.sin(ny*r + φy)

  if ktype == str('trefoil'): # Trefoil knot
    arc = arclength(nx,ny,nz, φx,φy,φz, ktype)
    intarc = int(round(arc,0))
    npts = 20*intarc
    t = np.linspace(0, 2*np.pi, npts)
    x = np.sin(t) + 2*np.sin(2*t)
    y = np.cos(t) - 2*np.cos(2*t)
    z = -np.sin(3*t)

  if ktype == str('torus'): # Torus knot - q times winded around the torus
    arc = arclength(nx,ny,nz, φx,φy,φz, ktype,q)
    intarc = int(round(arc,0))
    npts = 20*intarc
    t = np.linspace(0, 2*np.pi, npts)
    x = np.cos(t)*(1 + 0.5*np.cos(q*t))
    y = np.sin(t)*(1 + 0.5*np.cos(q*t))
    z = 0.5*np.sin(q*t)

  # points = np.hstack((np.vstack(np.round(x,8)), np.vstack(np.round(y,8)), np.vstack(np.round(z,8))))

  #Knot
  knot = [go.Scatter3d(x=x, y=y, z=z, mode='lines', line=dict(width=6, color='black'))]
  # knot = [go.Scatter3d(x=x, y=y, z=z, mode='lines+markers', marker=dict(size=2), line=dict(width=4, color='black'))]
  
  if nx==3 and ny==5 and nz==9 and φx==3 and φy==5 and φz==9:
    r = [(29/135)*np.pi,(155/135)*np.pi,  # +27
         (56/135)*np.pi,(128/135)*np.pi,  # +27
         (83/135)*np.pi,(101/135)*np.pi,  # +27
         (110/135)*np.pi,(74/135)*np.pi,  # +54
         (164/135)*np.pi,(20/135)*np.pi,  # +27
         (191/135)*np.pi,(-7/135)*np.pi,  # +27
         (218/135)*np.pi,(-34/135)*np.pi, # +27
         (245/135)*np.pi,(-61/135)*np.pi]
    # tot = 100
    # for i in range(tot):
    #   r.append((i/tot)*np.pi)
    #   r.append( np.pi-(i/tot)*np.pi) # horizontal (use tot/2)
    #   r.append(-(i/tot)*np.pi)
    #   r.append(-np.pi+(i/tot)*np.pi) # horizontal
  r = np.array(r)
  a,b,c = np.cos(nx*r + φx), np.cos(ny*r + φy), np.cos(nz*r + φz) # Nodes

  #Indexes for points opposed in vertical lines
  lines = [[2*i, 2*i+1] for i in range(len(r)//2)]

  #Bridges between antipodes
  for l in lines:
      knot.append(go.Scatter3d(x=[ a[l[0]], a[l[1]] ],
                               y=[ b[l[0]], b[l[1]] ],
                               z=[ c[l[0]], c[l[1]] ], line=dict(width=5), marker=dict(size=8)))
  fig = go.Figure(data=knot)
  camera = dict( up=dict(x=0, y=0, z=1), eye=dict(x=xx, y=yy, z=zz) )
  fig.update_layout(
                    title=f"nx={nx}, ny={ny}, nz={nz}<br>ϕx = ({φx/np.pi:.2f})π<br>ϕy = ({φy/np.pi:.2f})π<br>ϕz = ({φz/np.pi:.2f})π", 
                    # title=f"nx={nx}, ny={ny}, nz={nz}<br>ϕx = {φx:.0f}<br>ϕy = {φy:.0f}<br>ϕz = ({50*φz/np.pi:.0f}/50)π", 
                    scene_camera=camera, template='simple_white', width=500,height=500, # width=700,height=700, 
                    # paper_bgcolor="rgba(255,255,255,0.5)", # Background color and transparency
                    margin=dict(l=0, r=0, t=65, b=0), legend=dict(x=0.805,y=0.85
                                                                  ))
                    # legend=dict(x=0.805,y=1.11)) # Fix legend to keep plot centered within each frame
                    # scene=dict(xaxis=dict(range=[1, -1])))#, yaxis=dict(range=[1, -1]))) # Invert x- and/or y-axis
  # Save or display image
  if save_path:
      fig.write_image(save_path)
  else:
      fig.show()

  # Show circle
  if circle:
    xc, yc = np.cos(r), np.sin(r)
    circle = [go.Scatter(x=np.cos(t), y=np.sin(t), line=dict(width=2, color='black'))]

    #Circle bars
    for m in lines:
        circle.append( go.Scatter( x = [ xc[m[0]], xc[m[1]] ], y = [ yc[m[0]], yc[m[1]] ],
                                  line = dict(width=2), marker=dict(size=7)) )
    # Set up angle parameters
    angle_step = (27/135)*np.pi
    start_angle= ( 2/135)*np.pi
    angles = np.arange(start_angle, 2*np.pi, angle_step)

    # Add angle labels
    label_x = np.cos(angles)
    label_y = np.sin(angles)
    labels = [f"{angle:.3f}π" for angle in angles / np.pi]  # Express in terms of π
    circle.append( go.Scatter( x=label_x*1.2, y=label_y*1.08,  # Slightly outside the circle
        mode='text', text=labels, textposition='middle center', textfont=dict(size=10, color='black'), showlegend=False ))

    # Define two specific angles
    special_angles = np.array([ (2/135)*np.pi, (137/135)*np.pi ])
    marker_x = np.cos(special_angles)
    marker_y = np.sin(special_angles)
    # Add a black marker at those two positions
    circle.append(go.Scatter(x=marker_x,y=marker_y, mode='markers', marker=dict(size=6, color='black'), showlegend=False))

    fig = go.Figure(data=circle)
    fig.update_yaxes(scaleanchor = 'x', scaleratio = 1)
    fig.update_layout(template='simple_white', width=500,height=400,
                      margin=dict(l=5, r=95, t=5, b=5), legend=dict(x=1.0,y=0.97), # Fix legend to keep plot centered within each frame
                      # paper_bgcolor="rgba(255,255,255,0.0)", # Background color and transparency
                      # plot_bgcolor='rgba(255, 255, 255, 0)', 
                      )
    fig.show()

    # Show arc-length and npts
    print('Arc-length:', arc)
    print('Number of points:', npts)
  else:
    None
```

A Lissajous knot $K(t)$ is parameterized by

$$ x(t) = \cos(n_x t + \phi_x), $$

$$ y(t) = \cos(n_y t + \phi_y), $$

$$ z(t) = \cos(n_z t + \phi_z), $$

where $t,\ \phi_i \in [0, 2\pi]$, while the frequencies $n_i \in \mathbb{Z}$ are relatively prime, for $i=x,y,z.$

```python
# Arc-length function

from scipy.integrate import quad
import numpy as np

def arclength(nx,ny,nz, φx,φy,φz, ktype=str('Lissajous'), q=1):

  if ktype == str('Lissajous'): # Lissajous knot
    a = lambda t: np.sqrt((nx*np.sin(nx*t + φx))**2 + (ny*np.sin(ny*t + φy))**2 + (nz*np.sin(nz*t + φz))**2)

  elif ktype == str('fib'): # Fibonacci knot
    φx = φy = φz = 0.5
    a = lambda t: np.sqrt((nx*np.sin(nx*t + φx))**2 + (ny*np.sin(ny*t + φy))**2 +
                          (0.5*nz*np.sin(nz*t + φz) + 0.5*ny*np.cos(ny*t + φy))**2)

  elif ktype == str('trefoil'): # Trefoil knot
    a = lambda t: np.sqrt(  (  np.cos(t) + 4*np.cos(2*t))**2
                          + (- np.sin(t) + 4*np.sin(2*t))**2
                          + (3*np.cos(3*t))**2 )

  #The (2,3)-torus knot is also a trefoil knot.
  elif ktype == str('torus'): # Torus knot
    a = lambda t: np.sqrt(  (np.sin(t)*(1 + 0.5*np.cos(q*t)) + 0.5*q*np.sin(q*t)*np.cos(t))**2
                          + (np.cos(t)*(1 + 0.5*np.cos(q*t)) - 0.5*q*np.sin(q*t)*np.sin(t))**2
                          + (0.5*q*np.cos(q*t))**2 )

  arc = ( round( quad( a, 0, (1/2)*np.pi )[0], 2 ) + round( quad( a, (1/2)*np.pi, np.pi)[0], 2 )
      + round( quad( a, np.pi, (3/2)*np.pi)[0], 2 ) + round( quad( a, (3/2)*np.pi, 2*np.pi)[0], 2) )

  # print('The length of the', ktype, 'knot is:')
  return arc
```

Christoph Lamm argued that in order to __avoid self-intersections__, for fixed $\phi_z = 0$, the phases should __NOT__ be of some of the next three forms:

\begin{equation*}
    \phi_x = \dfrac{m}{n_z}\pi,
    \tag{1}
\end{equation*}

\begin{equation*}
    \phi_y = \dfrac{m}{n_z}\pi,
    \tag{2}
\end{equation*}

\begin{equation*}
    \phi_x = \frac{n_x}{n_y}\phi_y + \dfrac{m}{n_y}\pi,
    \tag{3}
\end{equation*}

where $m \in \mathbb{Z}$ such that $0 < m < n_z$ (Lamm [1997] "There are infinitely many Lissajous knots").

```python
# Lamm condition (3)

import numpy as np
import plotly.graph_objects as go

# def Liss(nx,ny,nz, φx=0,φy=0,φz=0, xx=-2.5,yy=0,zz=0, div=1, ktype=str('liss'), q=1): #(xx,yy,zz) is the camera position
#     r = []

def Liss(p,q,r,s,t,u,v,w, nx,ny,nz, φx=0,φy=0,φz=0, xx=-2.5,yy=0,zz=0, ktype=str('liss')): #(xx,yy,zz) is the camera position
    r = [p,q,r,s,t,u,v,w]

    # pi/3 Symmetry
    # 3,4,7, 0,0,np.pi/3
    # if φz == np.pi/3:
    #   r = [(5/12)*np.pi, (11/12)*np.pi, (17/12)*np.pi, (23/12)*np.pi, (2/12)*np.pi, (14/12)*np.pi]

    div=1
    if φx==0 and φy==0:
      div = nz
    elif φx==0 and φz==0:
      div = ny
    elif φy==0 and φz==0:
      div = nx
    for i in range(1,div): # Vertical Symmetry
      r.append((i/div)*np.pi)
      r.append(-(i/div)*np.pi)
      # print((i/div))

    if nx % 2 == 0:
      if φz == 0.5*np.pi: # Extended Vertical Symmetry
        r.append( 1/(2*ny)*np.pi )
        r.append( (2*ny-1)/(2*ny)*np.pi )
        r.append( -1/(2*ny)*np.pi )
        r.append( -(2*ny-1)/(2*ny)*np.pi )

      if φy == 0.5*np.pi and φx==0 and φz==0: # Horizontal Symmetry
        for i in range(1,nz,2):
          r.append( i/(2*nz)*np.pi )
          r.append( (2*nz-i)/(2*nz)*np.pi )
          r.append( -i/(2*nz)*np.pi )
          r.append( -(2*nz-i)/(2*nz)*np.pi )

    # Diagonal Symmetry
    if nx % 2 == 0:
      if φy == 0.2*np.pi:
        r.append( 1/(2*nz)*np.pi )
        r.append( -(2*nz-1)/(2*nz)*np.pi )

      if φy == 0.8*np.pi:
        r.append( -1/(2*nz)*np.pi )
        r.append( (2*nz-1)/(2*nz)*np.pi )

      if φy == 0.4*np.pi:
        r.append( -3/(2*nz)*np.pi )
        r.append( (2*nz-3)/(2*nz)*np.pi )

      if φy == 0.6*np.pi:
        r.append( 3/(2*nz)*np.pi )
        r.append( -(2*nz-3)/(2*nz)*np.pi )

    # Odd nx
    # if nx % 2 == 1:
    #   if φy == 0.2*np.pi:
    #     r.append( -3/(2*nz)*np.pi )
    #     r.append( (2*nz-3)/(2*nz)*np.pi )

    if ktype == str('liss'): # Lissajous knot
      arc = arclength(nx,ny,nz, φx,φy,φz)
      intarc = int(round(arc,0))
      npts = 20*intarc
      t = np.linspace(0, 2*np.pi, npts)
      x,y,z = np.cos(nx*t + φx), np.cos(ny*t + φy), np.cos(nz*t + φz) #Knot Points

    if ktype == str('fib'): # Fibonacci knot
      arc = arclength(nx,ny,nz, φx,φy,φz, ktype)
      intarc = int(round(arc,0))
      npts = 20*intarc
      t = np.linspace(0, 2*np.pi, npts)
      x,y = np.cos(nx*t + φx), np.cos(ny*t + φy)
      z = 0.5*np.cos(nz*t + φz) + 0.5*np.sin(ny*t + φy)
      r = np.array(r)
      a,b = np.cos(nx*r + φx), np.cos(ny*r + φy) # Fibonacci Nodes
      c = 0.5*np.cos(nz*r + φz) + 0.5*np.sin(ny*r + φy)

    if ktype == str('trefoil'): # Trefoil knot
      arc = arclength(nx,ny,nz, φx,φy,φz, ktype)
      intarc = int(round(arc,0))
      npts = 20*intarc
      t = np.linspace(0, 2*np.pi, npts)
      x = np.sin(t) + 2*np.sin(2*t)
      y = np.cos(t) - 2*np.cos(2*t)
      z = -np.sin(3*t)

    if ktype == str('torus'): # Torus knot - q times winded around the torus
      arc = arclength(nx,ny,nz, φx,φy,φz, ktype,q=1)
      intarc = int(round(arc,0))
      npts = 20*intarc
      t = np.linspace(0, 2*np.pi, npts)
      x = np.cos(t)*(1 + 0.5*np.cos(q*t))
      y = np.sin(t)*(1 + 0.5*np.cos(q*t))
      z = 0.5*np.sin(q*t)

    points = np.hstack((np.vstack(np.round(x,8)), np.vstack(np.round(y,8)), np.vstack(np.round(z,8))))

    #Knot
    knot = [go.Scatter3d(x=x, y=y, z=z, mode='lines', line=dict(width=4, color='black'))]

    r = np.array(r)
    a,b,c = np.cos(nx*r + φx), np.cos(ny*r + φy), np.cos(nz*r + φz) # Nodes

    #Indexes for points opposed in vertical lines
    lines = [[2*i, 2*i+1] for i in range(len(r)//2)]

    #Bridges between antipodes
    for l in lines:
        knot.append(go.Scatter3d(x=[ a[l[0]], a[l[1]] ],
                                 y=[ b[l[0]], b[l[1]] ],
                                 z=[ c[l[0]], c[l[1]] ], line=dict(width=4), marker=dict(size=5)))
    fig = go.Figure(data=knot)
    camera = dict( up=dict(x=0, y=0, z=1), eye=dict(x=xx, y=yy, z=zz) )
    fig.update_layout(scene_camera = camera, template='simple_white')
    fig.show()

    #Circle
    xc, yc = np.cos(r), np.sin(r)
    circle = [go.Scatter(x=np.cos(t), y=np.sin(t), line=dict(width=2, color='black'))]
    #Circle bars
    for m in lines:
        circle.append( go.Scatter( x = [ xc[m[0]], xc[m[1]] ], y = [ yc[m[0]], yc[m[1]] ],
                                   line = dict(width=2), marker=dict(size=7)) )
    fig = go.Figure(data=circle)
    fig.update_yaxes(scaleanchor = 'x', scaleratio = 1)
    fig.update_layout(template='simple_white', width=500,
                      xaxis = dict(dtick = 0.5))
    fig.show()

    print('Arc length:', arc)
    print('Number of points:', npts)
```

On the contrary, there are many cases for which we obtain __self-intersections__.

We are interested in knowing _a priori_ the number of self-intersections, because each self-intersection increases by one the genus of the surface of the knot, once it is thickened.

One particular case in which we can know the number of self-intersections is when only one coordinate $i = x,y,z$ has a phase $\boxed{\phi_i = \phi,}$ but $\phi$ equal to certain fractions of $\pi$ gives additional self-intersections.

Given that $\cos(-t) = \cos(t),$ it turns out that $(n_i-1)$ self-intersections are found for

\begin{equation}
\boxed{t = \pm\left(\dfrac{m}{n_i}\right)\pi,} \qquad m \in \mathbb{Z},\quad 0 < m < n_i,
\tag{4}
\end{equation}

at the coordinate
\begin{equation}
    \boxed{i = \pm \cos\phi_i,}
\end{equation}

with $i=x,y,z,$ and where the plus sign is for even $m$ and the minus sign for odd $m.$

This can be seen as a __vertical symmetry__ for the unit circle $S^1$ parameterized by $(\cos t, \sin t).$

Let's look at an example with the frequencies

$$\boxed{n_x = 2, n_y = 3, n_z = 5.}$$

We're going to see what happens if we have just one phase, first $\phi_x,$ then $\phi_y,$ and finally $\phi_z.$

### $$\color{orange}{\textbf{1. }}\ \boxed{\phi_x = \phi}$$

Since $n_x=2$ and $0 < m < n_x,$ this gives $m=1$ self-intersection. Susbstituting in $(4)$, we know that the self-intersection is given by

$$t = \pm \left(\dfrac{\pi}{2}\right)$$

at

$$\left( \cos\left(\pm \frac{2\pi}{2} + \phi\right),\ \cos \left( \pm \dfrac{3\pi}{2} \right),\ \cos \left( \pm \dfrac{5\pi}{2} \right) \right) = \big( -\cos\phi, 0, 0 \big),$$

where $$\cos(\pm \pi + \phi) = \cos(\pm\pi)\cos\phi - \sin(\pm\pi)\sin\phi = -\cos\phi.$$

```python
Liss(2,3,5, 4*np.pi/5,0,0)
```

### $$\color{orange}{\textbf{2. }} \boxed{\phi_y = \phi}$$

Now we get 2 self-intersections for $0 < m < n_y = 3,$ i.e. $m=1,2,$ for

$$t = \pm \bigg(\dfrac{\pi}{3}\bigg),\ \pm \left(\dfrac{2\pi}{3}\right) \quad \text{at} \quad \left( -\dfrac{1}{2}, -\cos\phi, \dfrac{1}{2} \right),\ \left( -\dfrac{1}{2}, \cos\phi, -\dfrac{1}{2} \right).$$

```python
Liss(2,3,5, 0,np.pi/3,0)
```

### $$\color{orange}{\textbf{3. }} \boxed{\phi_z = \phi}$$

Finally we get 4 self-intersections for $0 < m < n_z = 5,$ i.e. $m=1,2,3,4$ for

$$t = \pm \left(\dfrac{\pi}{5}\right),\ \pm \left(\dfrac{2\pi}{5}\right),\ \pm \left(\dfrac{3\pi}{5}\right),\ \pm \left(\dfrac{4\pi}{5}\right),$$
at
$$\left( 0.31, -0.31, -\cos\phi \right),\ \left( -0.81, -0.81, \cos\phi \right),\ \left( -0.81, 0.81, -\cos\phi \right),\ \left( 0.31, 0.31, \cos\phi \right).$$
where
$$\dfrac{\sqrt{5}-1}{4} \approx 0.31 \quad \text{y} \quad \dfrac{\sqrt{5}+1}{4} \approx 0.81.$$

```python
Liss(2,3,5, 0,0,np.pi/4)
```

### $$\color{orange}{\textbf{4. }} \boxed{\phi = \dfrac{m\pi}{n_i}}$$

There are some values for the phase that result in more intersections, these values are fractions of $\pi$ and result in __horizontal symmetries__  or __diagonal symmetries__ w.r.t. $S^1,$ rather than vertical symmetries.

$ $

For instance, $\phi_x = \dfrac{\pi}{3},\ \dfrac{2\pi}{3}$ produce 3 more and $\phi_x = \dfrac{\pi}{5},\ \dfrac{2\pi}{5},\ \dfrac{3\pi}{5},\ \dfrac{4\pi}{5}$ produce 2 more,

where the phases $\dfrac{\pi}{5}$ and $\dfrac{4\pi}{5}$ are inverses, just as $\dfrac{2\pi}{5}$ and $\dfrac{3\pi}{5}$.

$ $

Next, $\phi_y = \dfrac{\pi}{2}$ produces 4 more, while $\phi_y = \dfrac{\pi}{5},\ \dfrac{2\pi}{5},\ \dfrac{3\pi}{5},\ \dfrac{4\pi}{5}$ produce 1 more.

$ $

And $\phi_z = \dfrac{\pi}{2}$ produces 4 more, while $\phi_z = \dfrac{\pi}{3},\ \dfrac{2\pi}{3}$ produce 1 more.

```python
Liss(2,3,5, np.pi/2,0,np.pi/2)
```

## Diagonal Symmetry

```python
# Arc-length function

from scipy.integrate import quad
import numpy as np

def arclength(nx,ny,nz, φx,φy,φz, ktype=str('Lissajous'), q=1):

  if ktype == str('Lissajous'): # Lissajous knot
    a = lambda t: np.sqrt((nx*np.sin(nx*t + φx))**2 + (ny*np.sin(ny*t + φy))**2 + (nz*np.sin(nz*t + φz))**2)

  elif ktype == str('fib'): # Fibonacci knot
    φx = φy = φz = 0.5
    a = lambda t: np.sqrt((nx*np.sin(nx*t + φx))**2 + (ny*np.sin(ny*t + φy))**2 +
                          (0.5*nz*np.sin(nz*t + φz) + 0.5*ny*np.cos(ny*t + φy))**2)

  elif ktype == str('trefoil'): # Trefoil knot
    a = lambda t: np.sqrt(  (  np.cos(t) + 4*np.cos(2*t))**2
                          + (- np.sin(t) + 4*np.sin(2*t))**2
                          + (3*np.cos(3*t))**2 )

  #The (2,3)-torus knot is also a trefoil knot.
  elif ktype == str('torus'): # Torus knot
    a = lambda t: np.sqrt(  (np.sin(t)*(1 + 0.5*np.cos(q*t)) + 0.5*q*np.sin(q*t)*np.cos(t))**2
                          + (np.cos(t)*(1 + 0.5*np.cos(q*t)) - 0.5*q*np.sin(q*t)*np.sin(t))**2
                          + (0.5*q*np.cos(q*t))**2 )

  arc = ( round( quad( a, 0, (1/2)*np.pi )[0], 2 ) + round( quad( a, (1/2)*np.pi, np.pi)[0], 2 )
      + round( quad( a, np.pi, (3/2)*np.pi)[0], 2 ) + round( quad( a, (3/2)*np.pi, 2*np.pi)[0], 2) )

  # print('The length of the', ktype, 'knot is:')
  return arc
```

```python
# Plotting function for the Lissajous Knots

import numpy as np
import plotly.graph_objects as go

def pi_3(nx,ny,nz, φx=0,φy=0,φz=0, xx=-2.2,yy=0,zz=0, div=1): #(xx,yy,zz) is the camera position

    r = []

    # pi/3 Symmetry
    # 3,4,7, 0,0,np.pi/3
    # if φz == np.pi/3:
    #   r = [-(1/12)*np.pi, -(7/12)*np.pi,  (2/12)*np.pi, -(10/12)*np.pi,  (5/12)*np.pi, -(13/12)*np.pi]

    # Vertical Symmetry
    if φx==0 and φy==0:
      div = nz
    elif φx==0 and φz==0:
      div = ny
    elif φy==0 and φz==0:
      div = nx
    for i in range(1,div):
      r.append((i/div)*np.pi)
      r.append(-(i/div)*np.pi)
      # print((i/div))

    # Horizontal Symmetry
    if nx % 2 == 0:
      if φz == 0.5*np.pi:
        r.append( 1/(2*ny)*np.pi )
        r.append( (2*ny-1)/(2*ny)*np.pi )
        r.append( -1/(2*ny)*np.pi )
        r.append( -(2*ny-1)/(2*ny)*np.pi )

      if φy == 0.5*np.pi:
        for i in range(1,nz,2):
          r.append( i/(2*nz)*np.pi )
          r.append( (2*nz-i)/(2*nz)*np.pi )
          r.append( -i/(2*nz)*np.pi )
          r.append( -(2*nz-i)/(2*nz)*np.pi )

    # n*(pi/5) Diagonal Symmetry
    if nx % 2 == 0:
      if φy == (1/5)*np.pi:
        r.append( 1/(2*nz)*np.pi )
        r.append( -((2*nz-1)/(2*nz))*np.pi )

      if φy == (4/5)*np.pi:
        r.append( -1/(2*nz)*np.pi )
        r.append( ((2*nz-1)/(2*nz))*np.pi )

      if φy == (2/5)*np.pi:
        r.append( -3/(2*nz)*np.pi )
        r.append( ((2*nz-3)/(2*nz))*np.pi )

      if φy == (3/5)*np.pi:
        r.append( 3/(2*nz)*np.pi )
        r.append( -((2*nz-3)/(2*nz))*np.pi )


      if φx == (1/5)*np.pi:
        r.append( 1/(ny*nz)*np.pi )
        r.append( 11/(ny*nz)*np.pi )
        r.append( 1/(ny*nz)*np.pi - np.pi )
        r.append( 11/(ny*nz)*np.pi - np.pi )

    # n*(pi/3) Diagonal Symmetry

    # r.append((2/15)*np.pi)
    # r.append((8/15)*np.pi)
    # r.append((-1/15)*np.pi)
    # r.append((11/15)*np.pi)
    # r.append((-4/15)*np.pi)
    # r.append((14/15)*np.pi)
    # r.append((-7/15)*np.pi)
    # r.append((17/15)*np.pi)

    if φx == (1/3)*np.pi:
      for i in range(-1,ny):
        r.append( - np.pi/(ny*nz)  - i*ny*np.pi/(ny*nz) )
        r.append( 11*np.pi/(ny*nz) + i*ny*np.pi/(ny*nz) )

    if nx==2 and ny==3 and nz==5 and φz == (1/3)*np.pi:
      r.append( -(1/(nx*ny))*np.pi )
      r.append( (5/(nx*ny))*np.pi )

    if nx==3 and ny==4 and nz==7 and φz == (1/3)*np.pi:
      for i in range(0,nx):
        r.append( -np.pi/(nx*ny)         + i*nx*np.pi/(nx*ny) )
        r.append( -np.pi/(nx*ny)-np.pi/2 - i*nx*np.pi/(nx*ny) )
          

    r = np.array(r)
    a,b,c = np.cos(nx*r + φx), np.cos(ny*r + φy), np.cos(nz*r + φz) #Nodes

    arc = arclength(nx,ny,nz, φx,φy,φz)
    intarc = int(round(arc,0))
    npts = 10*intarc
    t = np.linspace(0, 2*np.pi, npts)
    x,y,z = np.cos(nx*t + φx), np.cos(ny*t + φy), np.cos(nz*t + φz) #Knot Points

    points = np.hstack((np.vstack(np.round(x,8)), np.vstack(np.round(y,8)), np.vstack(np.round(z,8))))

    #Knot
    knot = [go.Scatter3d(x=x, y=y, z=z, mode='lines', line=dict(width=4, color='black'))]

    #Indexes for points opposed in vertical lines
    lines = [[2*i, 2*i+1] for i in range(len(r)//2)]

    #Bridges between antipodes
    for l in lines:
        knot.append(go.Scatter3d(x=[ a[l[0]], a[l[1]] ],
                                 y=[ b[l[0]], b[l[1]] ],
                                 z=[ c[l[0]], c[l[1]] ], line=dict(width=4), marker=dict(size=5)))
    fig = go.Figure(data=knot)
    camera = dict( up=dict(x=0, y=0, z=1), eye=dict(x=xx, y=yy, z=zz) )
    fig.update_layout(scene_camera = camera, template='simple_white')
    fig.show()

    #Circle
    xc, yc = np.cos(r), np.sin(r)
    circle = [go.Scatter(x=np.cos(t), y=np.sin(t), line=dict(width=2, color='black'))]
    #Circle bars
    for m in lines:
        circle.append( go.Scatter( x = [ xc[m[0]], xc[m[1]] ], y = [ yc[m[0]], yc[m[1]] ],
                                   line = dict(width=2), marker=dict(size=7)) )
    fig = go.Figure(data=circle)
    fig.update_yaxes(scaleanchor = 'x', scaleratio = 1)
    fig.update_layout(template='simple_white', width=500,
                      xaxis = dict(dtick = 0.5))
    fig.show()

    print('Arc length:', arc)
    print('Number of points:', npts)
```

```python
# Plotting function for n/20 Symmetry

import numpy as np
import plotly.graph_objects as go

def Symm20(nx,ny,nz, φx=0,φy=0,φz=0, a=0,b=0,c=0,d=0,e=0,f=0, xx=-2.5,yy=0,zz=0, div=1): #(xx,yy,zz) is the camera position

    r = []

    # 3,5,8, 0,np.pi/3,0
    r.append(a), r.append(b), r.append(c), r.append(d), r.append(e), r.append(f)

    # pi/3 Symmetry
    # 3,4,7, 0,0,np.pi/3
    if φz == np.pi/3:
      r = [(5/12)*np.pi, (11/12)*np.pi, (17/12)*np.pi, (23/12)*np.pi, (2/12)*np.pi, (14/12)*np.pi]

    if φx==0 and φy==0:
      div = nz
    elif φx==0 and φz==0:
      div = ny
    elif φy==0 and φz==0:
      div = nx
    for i in range(1,div): # Vertical Symmetry
      r.append((i/div)*np.pi)
      r.append(-(i/div)*np.pi)
      # print((i/div))

    if φz == 0.5*np.pi: # Extended Vertical Symmetry
      r.append( 1/(2*ny)*np.pi )
      r.append( (2*ny-1)/(2*ny)*np.pi )
      r.append( -1/(2*ny)*np.pi )
      r.append( -(2*ny-1)/(2*ny)*np.pi )

    if nx % 2 == 0:
      if φy == 0.5*np.pi: # Horizontal Symmetry
        for i in range(1,nz,2):
          r.append( i/(2*nz)*np.pi )
          r.append( (2*nz-i)/(2*nz)*np.pi )
          r.append( -i/(2*nz)*np.pi )
          r.append( -(2*nz-i)/(2*nz)*np.pi )

    # Diagonal Symmetry
    if nx % 2 == 0:
      if φy == 0.2*np.pi:
        r.append( 1/(2*nz)*np.pi )
        r.append( -(2*nz-1)/(2*nz)*np.pi )

      if φy == 0.8*np.pi:
        r.append( -1/(2*nz)*np.pi )
        r.append( (2*nz-1)/(2*nz)*np.pi )

      if φy == 0.4*np.pi:
        r.append( -3/(2*nz)*np.pi )
        r.append( (2*nz-3)/(2*nz)*np.pi )

      if φy == 0.6*np.pi:
        r.append( 3/(2*nz)*np.pi )
        r.append( -(2*nz-3)/(2*nz)*np.pi )

    # Odd nx
    # if nx % 2 == 1:
    #   if φy == 0.2*np.pi:
    #     r.append( -3/(2*nz)*np.pi )
    #     r.append( (2*nz-3)/(2*nz)*np.pi )

    r = np.array(r)
    a,b,c = np.cos(nx*r + φx), np.cos(ny*r + φy), np.cos(nz*r + φz) #Nodes

    arc = arclength(nx,ny,nz, φx,φy,φz)
    intarc = int(round(arc,0))
    npts = 10*intarc
    t = np.linspace(0, 2*np.pi, npts)
    x,y,z = np.cos(nx*t + φx), np.cos(ny*t + φy), np.cos(nz*t + φz) #Knot Points

    points = np.hstack((np.vstack(np.round(x,8)), np.vstack(np.round(y,8)), np.vstack(np.round(z,8))))

    #Knot
    knot = [go.Scatter3d(x=x, y=y, z=z, mode='lines', line=dict(width=4, color='black'))]

    #Indexes for points opposed in vertical lines
    lines = [[2*i, 2*i+1] for i in range(len(r)//2)]

    #Bridges between antipodes
    for l in lines:
        knot.append(go.Scatter3d(x=[ a[l[0]], a[l[1]] ],
                                 y=[ b[l[0]], b[l[1]] ],
                                 z=[ c[l[0]], c[l[1]] ], line=dict(width=4), marker=dict(size=5)))
    fig = go.Figure(data=knot)
    camera = dict( up=dict(x=0, y=0, z=1), eye=dict(x=xx, y=yy, z=zz) )
    fig.update_layout(scene_camera = camera, template='simple_white')
    fig.show()

    #Circle
    xc, yc = np.cos(r), np.sin(r)
    circle = [go.Scatter(x=np.cos(t), y=np.sin(t), line=dict(width=2, color='black'))]
    #Circle bars
    for m in lines:
        circle.append( go.Scatter( x = [ xc[m[0]], xc[m[1]] ], y = [ yc[m[0]], yc[m[1]] ],
                                   line = dict(width=2), marker=dict(size=7)) )
    fig = go.Figure(data=circle)
    fig.update_yaxes(scaleanchor = 'x', scaleratio = 1)
    fig.update_layout(template='simple_white', width=500,
                      xaxis = dict(dtick = 0.5))
    fig.show()

    print('Arc length:', arc)
    print('Number of points:', npts)
```
